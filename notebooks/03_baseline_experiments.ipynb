Implements TF-IDF + Logistic Regression / SVM baselines for classification (Sentiment, Intent) and CRF for NER.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

from src.models.baselines import TFIDFClassifier

train_texts = ['Acha movie hai', 'Bohot bekar tha', 'Normal film', 'Worst experience ever', 'Great acting']
train_labels = ['positive', 'negative', 'neutral', 'negative', 'positive']

clf = TFIDFClassifier(classifier_type='svm')
clf.fit(train_texts, train_labels)
eval_res = clf.evaluate(['Acchi film thi', 'Khaana bekar tha'], ['positive', 'negative'])
print(eval_res['report'])
```
